```python
x = ['Sushil','Roshan','Lucifer','Devil']
y = [92 , 100 , 68 , 82]
plt.xlabel("Monsters",fontsize = 12)                 # fontsize increases size of text
plt.ylabel("Power Level",fontsize = 12)              # fontsize increases size of text
plt.title("RSP",fontsize = 18)                       # fontsize increases size of text
plt.bar(x,y,color = ['k','b','y','c'] , width=0.7 , edgecolor = 'r' , linewidth = 2 , label = "Monster")
plt.legend()
# in that you can also use align='edge' and aling = 'center'
# and linewidth increases edgecolor size
# you can also add linestyle
# by label legend() function is compesary
plt.show()
```

```python
# stacked bar graph
z = [28,45,34,73]
plt.bar(x,y,color ='r', label = "Power" , edgecolor = 'k')
plt.bar(x,z,color ='c', label = "Health" , edgecolor = 'k')
plt.legend()
plt.show()
```

```python
x = ['Sushil','Roshan','Lucifer','Devil']
y = [92 , 100 , 68 , 82]
z = [28,45,34,73]

p = np.arange(len(x))
width = 0.2
p1 = [j + width  for j in p]

plt.xlabel("Monsters",fontsize = 12)                 # fontsize increases size of text
plt.ylabel("Power Level",fontsize = 12)              # fontsize increases size of text
plt.title("RSP",fontsize = 18)  

plt.bar(p,y,color ='r', label = "Power" ,width=0.2, edgecolor = 'k')
plt.bar(p1,z,color ='c', label = "Health",width=0.2, edgecolor = 'k')

plt.xticks(p + width/2, x , rotation = 20)           # divided by 2 for names in center in bars on x
plt.yticks(rotation = 25)                            # by rotation parameters of x and y can be rotate
plt.legend()
plt.show()
```

# Scatter Plot

```python
import matplotlib.pyplot as plt
x = [2,4,6,5,4,2,3]
y = [1,4,2,6,8,4,7]
plt.title("scatter", fontsize = 18)
plt.xlabel("scatter - x", fontsize = 13)
plt.ylabel("scatter - y", fontsize = 13)
c = ['r','c','k','b','g','y','m']
sizes = [150,205,400,400,45,100,80]
plt.scatter(x,y , color=c , s = sizes,alpha = 0.5)          # alpha can blur braph it decrease transperency
plt.show()
```

# Step Plot

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5,6]
y = [1,2,3,4,5,6]
plt.step(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5,6]
y = [2,2,3,5,4,6]
plt.step(x,y, color="r", marker="o", ms = 10, mfc = "g", label = "Python")
# marker is a shape of point
# ms is a marker size, which help insrease size of a marker
# mfc is a marker face color, which help to change color of marker
plt.title("Python", fontsize=18)
plt.xlabel("X - axis", fontsize = 12)
plt.ylabel("Y - axis", fontsize = 12)
plt.legend(loc = 2)
plt.grid()
plt.show()
```

# Fill_Between Plot

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5,6]
y = [1,2,3,4,5,6]
plt.plot(x,y, color="r")
plt.fill_between(x,y)
plt.title("Roshan", fontsize=15)
plt.xlabel("X - axis", fontsize=12)
plt.ylabel("Y - axis", fontsize=12)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5,6]
y = [1,2,3,4,5,6]
plt.plot(x,y, color="r")
plt.fill_between(x=[2,4], y1=2, y2=5)
plt.title("Roshan", fontsize=15)
plt.xlabel("X - axis", fontsize=12)
plt.ylabel("Y - axis", fontsize=12)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1,2,3,4,5,6])
y = np.array([1,2,3,4,5,6])
plt.plot(x,y, color="r")
plt.fill_between(x,y, color="g", where=(x>=2) & (x<=4))
# '>=' not supported between instances of 'list' and 'int' so we use numpy array
# where parameter is for a range 
plt.title("Roshan", fontsize=15)
plt.xlabel("X - axis", fontsize=12)
plt.ylabel("Y - axis", fontsize=12)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1,2,3,4,5,6])
y = np.array([1,2,3,4,5,6])
plt.plot(x,y, color="r")
plt.fill_between(x,y, color="g", where=(x>=2) & (x<=4), alpha = 0.5)
# alpha use for to decrease the hardness of a color
plt.title("Roshan", fontsize=15)
plt.xlabel("X - axis", fontsize=12)
plt.ylabel("Y - axis", fontsize=12)
plt.show()
```

# Subplot

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5,6]
y = [1,2,3,4,5,6]
z = [20,40,50,40,10,60]
x1 = ["R","S","P","D","L"]
y1 = [50,48,49,40,35]
plt.subplot(2,2,1)                 # (2,2,1) here first 2 are a size, second 2 are a columns and 1 is a position
plt.plot(x,y, color="r")
plt.subplot(2,2,2)
plt.pie([1], colors="r")
plt.subplot(2,2,3)
plt.pie(x)
plt.subplot(2,2,4)
plt.bar(x1,y1)
plt.show()
```
