```python
x = [2, 4, 5, 8]
y = [1, 2, 1, 5]
plt.title("first plot")
plt.xlabel("X position")
plt.ylabel("Y position")
plt.plot(x, y, color='r', linewidth=2)
plt.show()
```

```python
year = [2020, 2021, 2022, 2023, 2024, 2025]
populations = [1000000, 1500000, 2500000, 4000000, 3500000, 5000000]
plt.title('Population per year')
plt.xlabel("Year")
plt.ylabel("Population")
plt.bar(year, populations, color='r', edgecolor='k', linewidth=1)
plt.show()
```

```python
x = ['Sushil','Roshan','Lucifer','Devil']
y = [92 , 100 , 68 , 82]
plt.xlabel("Monsters",fontsize = 12)      
plt.ylabel("Power Level",fontsize = 12)        
plt.title("RSP",fontsize = 18)                     
plt.bar(x,y,color = ['r','b','y','c'] , width=0.7 , edgecolor = 'k' , linewidth = 1 , label = "Monster")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [20,30,40,44]
y = ['Lucifer','Devil','Sushil','Roshan']
plt.pie(x,labels=y, autopct="%2.2f %%")         # autopct="%2.2f" for percentages
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [20,30,40,44]
y = ['Lucifer','Devil','Sushil','Roshan']
ex = [0.1,0.0,0.0,0.0]
plt.pie(x,labels=y, explode=ex, autopct="%.2f %%")          # explode help to represent data
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [20,30,40,44]
y = ['Lucifer','Devil','Sushil','Roshan']
ex = [0.1,0.0,0.0,0.0]
plt.pie(x,labels=y, explode=ex, autopct="%.2f %%", shadow= True)  
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [20,30,40,44]
y = ['Lucifer','Devil','Sushil','Roshan']
ex = [0.1,0.0,0.0,0.0]
plt.pie(x,labels=y, explode=ex, autopct="%.2f %%", shadow= True, wedgeprops={"linewidth": 2, "edgecolor": "w"})  
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [2,4,6,5,4,2,3]
y = [1,4,2,6,8,4,7]
plt.title("scatter", fontsize = 18)
plt.xlabel("scatter - x", fontsize = 13)
plt.ylabel("scatter - y", fontsize = 13)
c = ['r','c','k','b','g','y','m']
sizes = [150,205,400,400,45,100,80]
plt.scatter(x,y , color=c , s = sizes)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [2,4,6,5,4,2,3]
y = [1,4,2,6,8,4,7]
plt.title("scatter", fontsize = 18)
plt.xlabel("scatter - x", fontsize = 13)
plt.ylabel("scatter - y", fontsize = 13)
c = ['r','c','k','b','g','y','m']
sizes = [150,205,400,400,45,100,80]
plt.scatter(x,y , color=c , s = sizes, alpha = 0.5)          # alpha can blur braph it decrease transperency
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [2,4,6,5,4,2,3]
y = [1,4,2,6,8,4,7]
plt.title("scatter", fontsize = 18)
plt.xlabel("scatter - x", fontsize = 13)
plt.ylabel("scatter - y", fontsize = 13)
c = ['r','c','k','b','g','y','m']
sizes = [150,205,400,400,45,100,80]
plt.scatter(x,y , color=c , s = sizes, marker='*', edgecolors='k')     
plt.show()
```
