# K-Nearest Neighbors (KNN)
<b>Problem</b>
<p>A retail company wants to predict customer purchasing behavior based on their age, salary and past purchase history. The company aim to use K-Nearest Neighbors (KNN) algorithm to classify customers into potential buying groups to personalize marketing statergies. This predictive model will help the company understand and target specific customer segments more effectively, there by increasing sales and customer satisfaction</p>
<p>KNN Because of classification</p>

```python
# Step no. 1
import numpy as np
from  sklearn.model_selection import train_test_split
from sklearn.preprocessing import  StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Step no. 2
data = np.array([[25,50000,2], [30,80000,1], [35,60000,3], [20,30000,2], [40,90000,1], [45,75000,2]])
# first is age, second is salary and third is how many products customer buy
labels = np.array([1, 2, 1, 0, 2, 1])
# 0 is low, 1 is medium, 2 is high
```

```python
# Step no. 3
x_train, x_test, y_train, y_test = train_test_split(data, labels, test_size=0.2, random_state=42)
```

```python
# Step no. 4
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.fit_transform(x_test)
```

```python
# Step no. 5
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(x_train, y_train)
```

```python
accuracy = knn.score(x_test, y_test)
accuracy
```

# Naive Bayes
<b>Problem</b>
<p>A company is looking to develop a text classification model to categorize customer reviews into positive or negative sentiments. They want to use Naive Bayes classification to automatically analyze and classify the reviews they receive, aiming to understand customer satisfaction levels and sentiments. The company desires a model that can accurately predict whether a customer review expresses a positive and negative sentiments</p>

```python
# Step no. 1
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB      
''' The multinomial Naive Bayes classifier is suitable for classification
 with discrete features (e.g., word counts for text classification) '''
```

```python
# Step no. 2
reviews = ["The product is excellent and work perfectly",
           "The product is very bad and disappointing",
           "Terrible product and waste of money",
           "I love this product and work amazing"]
sentiments = np.array([1,0,0,1])
```

```python
# Step no. 3
vectorizer = CountVectorizer()
x = vectorizer.fit_transform(reviews)    
# fit_transform  is used to transform the reviews into a matrix of token counts (vectors).
```

```python
# Step no. 4
classifier = MultinomialNB()
classifier.fit(x,sentiments)
```

```python
# Step no. 5
def classify_new_review(new_review):
    review_vetctorize = vectorizer.transform([new_review])
    prediction = classifier.predict(review_vetctorize)
    if prediction[0] == 1:
        return "Positive Sentiment"
    else:
        return "Negative Sentiment"
```

# K means Clustering Algorithm
<b>Problem</b>
<p>A mall wants to understand its customer base better to improve its marketing strategy. The goal is to customers into groups based on their annual income and spending score, so the marketing team can tailor their campaigns to each cluster's preferences and behaviors</p>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import  KMeans
```

```python
# generate data
np.random.seed(42)  
data = {
    'annual_income' : np.random.randint(30000,100000,100),
    # 30000 is starting, 100000 is ending, 100 is limit
    'spending_score' : np.random.randint(1,100,100)
    # spending score is 1 to 100 and 100 is limit
}
df = pd.DataFrame(data)
df
```

```python
x = df['annual_income']
y = df['spending_score']
plt.scatter(x,y)
plt.title('annual income vs spending score', fontsize=14, color='blue')
plt.xlabel('annual income', fontsize=12, color='blue')
plt.ylabel('spending score', fontsize=12, color='blue')
plt.show()
```

```python
xm = df.values
Kmeans = KMeans(n_clusters=3, random_state=42)
df['cluster'] = Kmeans.fit_predict(xm)    # fit_predict add cluster
df
```

```python
plt.scatter(x,y,c = df['cluster'], cmap='rainbow')
plt.title('annual income vs spending score', fontsize=14, color='blue')
plt.xlabel('annual income', fontsize=12, color='blue')
plt.ylabel('spending score', fontsize=12, color='blue')
plt.show()
```

```python
# user input
user_input = {
    'annual_income' : [75000],
    'spending_score' : [70]
}
user_df = pd.DataFrame(user_input)
user_cluster = Kmeans.predict(user_df)
user_cluster
print(f'user belongs to cluster {user_cluster[0]}')
```

# DBSCAN Clustering
<b>Problem</b>
<p>Imagine your working for a e-commerce company and the marketing team wants to target customer more effectively by creating segments based on their purchasing behavior. However, traditional segmentation methods are not providing satisfactory result.</p>

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
data = np.random.rand(100,2)
data
```

```python
scalar = StandardScaler()
scalar_data = scalar.fit_transform(data)
scalar_data
```

```python
# apply DBSCAN
# eps = distance between two samples, 0.3 = 30%
# min_sample = 5, means 5 sample nearest from data point and data point is core point
dbscan = DBSCAN(eps=0.3, min_samples=5)
cluster = dbscan.fit_predict(scalar_data)
cluster
```

```python
# visualize
plt.scatter(data[:,0], data[:,1], c=cluster, cmap='viridis', marker='o', s=50)
plt.title("DBSCAN Clustering")
plt.xlabel("feature 1")
plt.ylabel("feature 2")
plt.show()
```

```python
result_df = pd.DataFrame(
    {
        'feature 1' : data[:,0],
        'feature 2' : data[:,1],
        'cluster' : cluster
    }
)
result_df
```

```python
print("number of clusters : ",len(np.unique(cluster)))
print("size of each cluster : ")
print(result_df['cluster'].value_counts())
```

```python
# user input
user_input = np.array([[0.6,0.7]])
user_input_scaled = scalar.transform(user_input)
user_cluster = dbscan.fit_predict(user_input_scaled)
print("user input belongs to cluster",user_cluster[0])
```

# Gaussian Mixture Model
<b>Problem</b>
<p>As a retail business owner, you want to understand and categorize your customers into distinct segments based on their purchasing behavior. This segmentation will help you tailor marketing strategies and promotions to different customer groups, ultimately maximizing sales and customer satisfaction.</p>

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import  StandardScaler
```

```python
np.random.seed(42)
# loc = mean, scale = standard deviation, size = number of sample 
segment_1 = np.random.normal(loc=30, scale=5, size=100)
segment_2 = np.random.normal(loc=60, scale=10, size=150)
segment_3 = np.random.normal(loc=90, scale=8, size=120)
data = np.concatenate([segment_1, segment_2, segment_3]).reshape(-1,1) 
# concatenate for combine 3 segments in a single data set
data
```

```python
scalar = StandardScaler()
scalar_data = scalar.fit_transform(data)
scalar_data
```

```python
gmm = GaussianMixture(n_components=3, random_state=42)
gmm.fit(scalar_data)
```

```python
cluster_label = gmm.predict(scalar_data)
cluster_label
```

```python
# visualize
plt.scatter(data, np.zeros_like(data), c=cluster_label, cmap='viridis')
plt.title("cluster segmentaion")
plt.xlabel("purchase amount")
plt.show()
```
